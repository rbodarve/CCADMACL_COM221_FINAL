```python
import re

import pandas as pd

targetCSV = "CSV/imdb_episodes.csv"
```

```python
df = pd.read_csv(targetCSV)
df
```

```python
df['Poster']
```

```python
# Modify the poster column to use snake case format instead of the normal one with spaces.

def snake_case(s):
	print(s)
	if str(s) == "nan":
		return s
	return re.sub(r'\s+', '_', s)

df['Poster'] = df['Poster'].apply(snake_case)
df
```

```python
df["Slug"].duplicated().sum()
```

```python
df = df[~df["Slug"].duplicated()]
df
```

```python
# Save back to the CSV file
df.to_csv(f"modified_{targetCSV}", index=False)
```
