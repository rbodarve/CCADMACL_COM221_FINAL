```python
from bs4 import BeautifulSoup
from urllib.parse import urlparse
import json
import os
import re
import requests
```

```python

headers = {
    'Accept': '*/*',
    'Connection': 'keep-alive',
    'User-Agent': 'Mozilla/5.0 (Windows NT 6.0) AppleWebKit/537.36 (KHTML,like Gecko) Chrome/70.0.3538.110 Safari/537.36', 
    'Accept-Language': 'en-US;q=0.5,en;q=0.3',
    'Cache-Control': 'max-age=0',
    'Upgrade-Insecure-Requests': '1'
}

image_path = 'imdb_episode_images'
slug = 'tt7707100'
```

```python
try:
	# Fetch the episode data
	response = requests.get(f'https://www.imdb.com/title/{slug}', headers=headers)
	soup = BeautifulSoup(response.content, 'html.parser')

	# Find the script tag containing the JSON data
	script = soup.find('script', {'type': "application/ld+json"})

	# Check if the script tag is found and contains data
	if script:
		data = json.loads(script.text)
		
  		# Generate the poster name
		img_url = data.get('image', "N/A")
		ogName = data.get('name', 'N/A')
		img_location = "N/A"
  
		ext = os.path.splitext(urlparse(img_url).path)[1]
		valid_filename = slug
		if not ext:
			ext = '.jpg'
  
		if img_url != "N/A":
			img_location = f"{image_path}\\{re.sub(r"\s+", "_", valid_filename)}{ext}"
  
		# Print a comma separated value string
		desc = re.sub(r"\s+", " ", str(data.get('description', 'N/A')).replace("&", "&amp;").replace("\"", "&quot;").replace("\'", "&apos;").replace("\n", " "))
		genre = str(data.get('genre', 'N/A'))
  
		if genre != 'N/A' and len(genre.strip()[1:-1].split(',')) > 1:
			genre = f"\"{genre}\""

		# Slug, Title, Rating, Rating Count, Date, Genres, Poster
		print(f"{slug},{ogName},{data.get('aggregateRating', {}).get('ratingValue', 'N/A')},{data.get('datePublished', 'N/A')},{genre},{img_location}")
except Exception as e:
		print(f"Error saving episode: {str(e)}")
```
