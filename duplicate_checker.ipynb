```python
import os

import numpy as np
import pandas as pd

dir = "imdb_episode_images"
targetDir = os.path.abspath(dir)
targetCSV = "CSV/imdb_episodes.csv"
col = "Slug"
```

```python
df = pd.read_csv(targetCSV)
df
```

# Actual Checking

## CSV Dupes

```python
dupCount = int(df[col].duplicated().sum())

if dupCount == 0:
	print(f"No duplicates found ({dupCount})")
else:
	print(f"There are {dupCount} duplicates found")
```

```python
df[df[col].duplicated()]
```

```python
noPosters = df["Poster"].isnull().sum()
print(f"There are {noPosters} rows with no posters")
```

```python
posterCount = len(df["Poster"]) - int(df["Poster"].isnull().sum())
print(f"Number of series that has poster: {posterCount}")
```

```python
dfCpy = df.copy()
dfCpy.dropna(subset=["Poster"], inplace=True)
dfCpy["Poster - CI"] = dfCpy["Poster"].str.upper()
dfDupPoster = dfCpy["Poster - CI"].duplicated().sum()

if dfDupPoster == 0:
	print(f"No duplicates found in Poster column ({dfDupPoster})")
else:
	print(f"There are {dfDupPoster} duplicates found in Poster column")
```

```python
dfCpy[dfCpy["Poster - CI"].duplicated()]
```

```python
dropDupes = False
"""
Determines if duplicates should be dropped. If this is set to `True`, the duplicates will be dropped and the CSV will be saved.
"""

if dropDupes:
    df.drop_duplicates(subset=["Poster"], inplace=True)
    df.to_csv(targetCSV, index=False)
```

## Image Dupes

```python
files = os.listdir(targetDir)
files.sort()
files = [f"{dir}\\{file}" for file in files if ".ipynb" not in file]
```

```python
imageList = np.array(files)
imageList
```

```python
imgDf = pd.DataFrame(imageList, columns=["Images"])
imgDf
```

```python
imgDf["Images - CI"] = imgDf["Images"].str.upper()
imgDupCount = int(imgDf["Images - CI"].duplicated().sum())

if imgDupCount == 0:
	print(f"No duplicates found ({imgDupCount})")
else:
	print(f"There are {imgDupCount} duplicates found")
```

# Check Missing Posters

```python
missingPosters = df[~df["Poster"].isin(imgDf["Images"])].copy()
missingPosters
```

```python
dropMissing = False

if dropMissing:
    df.drop(missingPosters.index, inplace=True)
    df.to_csv(targetCSV, index=False)
```
